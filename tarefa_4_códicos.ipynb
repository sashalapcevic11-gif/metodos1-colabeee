```python
#questao 1-a

import numpy as np

A=np.array([[10.,2.,-1.],[-3.,-5.,2.],[1.,1.,6.]])
B=np.array([[27.],[-61.5],[-21.5]])

Aum=np.hstack((A,B))
n=len(B)

for i in range(n-1):
  for j in range(i+1,n):
    fator= Aum[j,i]/Aum[i,i]
    Aum[j,i:n+1] = Aum[j,i:n+1] - fator*Aum[i,i:n+1]

print ('matriz aumentada:',Aum)

x = np.zeros(n)
x[n-1] = Aum[n-1,n] / Aum[n-1,n-1]
for i in range(n-2,-1,-1):
  soma = 0
  for j in range(i+1,n):
    soma += Aum[i,j]*x[j]

  x[i] = (Aum[i,n]-soma)/Aum[i,i]

print ('resultado:',x)

```

```python
#questao 2-a

import numpy as np

A=np.array([[2.,-6.,-1.],[-1.,-1.,7.],[-8.,1.,-2.]])
B=np.array([[-38.],[-34.],[-20.]])

Aum=np.hstack((A,B))
n=len(B)

for i in range(n-1):

  p = i + np.argmax(abs(Aum[i:n,i]))
  Aum[[i,p],:] = Aum[[p,i],:]
  for j in range(i+1,n):
    fator= Aum[j,i]/Aum[i,i]
    Aum[j,i:n+1] = Aum[j,i:n+1] - fator*Aum[i,i:n+1]

print ('matriz aumentada:',Aum)

x = np.zeros(n)
x[n-1] = Aum[n-1,n] / Aum[n-1,n-1]
for i in range(n-2,-1,-1):
  soma = 0
  for j in range(i+1,n):
    soma += Aum[i,j]*x[j]

  x[i] = (Aum[i,n]-soma)/Aum[i,i]

print ('resultado:',x)
```
